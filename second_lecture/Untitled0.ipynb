```python
import pandas as pd
data = pd.read_csv("CarPrice_Assignment.csv")
data.drop(["car_ID", "symboling"], axis=1, inplace=True)
data["CarName"] = data["CarName"].map(lambda x: x.split()[0])
data["CarName"] = data["CarName"].replace({"maxda":"mazda",
                                           "porcshce":"porsche",
                                           "toyouta":"toyota",
                                           "vokswagen":"volkswagen",
                                           "vw":"volkswagen"
                                           })
cat_columns = data.select_dtypes(include='object').columns
from sklearn.preprocessing import LabelEncoder
my_encoder = LabelEncoder()

for column in cat_columns:
  if len(data[column].unique()) <= 3:
         data[column] = my_encoder.fit_transform(data[column])
data = pd.get_dummies(data,columns = data.select_dtypes(include='object').columns,
                      drop_first=True, dtype=int)

data.head()
```

```python
import statsmodels.api as sm
y = data['price']
X = data.drop('price', axis=1)
X = sm.add_constant(X)
model = sm.OLS(endog=y, exog = X).fit()
print(model.summary())
```
